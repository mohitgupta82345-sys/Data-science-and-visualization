```python
## 1.	Create a NumPy array from [10, 20, 30, 40, 50]. Display its shape, size, and data type.
arr = np.array([10, 20 , 30 , 40, 50])
print(type(arr))
print(arr.shape)
print(arr.size)
arr.dtype
```

```python
                                                                  ## 2. Array Arithmetic
```

```python
## 6.	Given two NumPy arrays, perform element-wise addition, subtraction, multiplication, and division.
arr=np.array  ([1,2,3,4,5])
arr2=np.array([6,7,8,9,10])
print("sum:" ,arr+arr2)
print( "diff:" , arr-arr2)
print("multi:" , arr*arr2)
print("divide:" , arr/arr2)
```

```python
## 7.	Calculate the square and cube of every element in a NumPy array.
arr = np.array([1,2,3])
print(arr**2)
print(arr**3)
```

```python
## 8.	Increase every element of an array by 15%.
arr = np.array([1,2,3,4,5])
print(arr*1.15)
```

```python
##9.	Given two arrays, find their element-wise maximum and minimum.
arr = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
print(np.maximum(arr,arr2))
print(np.minimum(arr,arr2))
```

```python
## 10.	Create an array from 0 to 20 and extract the first five elements, last five elements, elements from index 5 to 15, and every second element.
arr = np.arange(0,21)
print(arr[0:5])
print(arr[16:21])
print(arr[5:16])
print(arr[::2])
```
